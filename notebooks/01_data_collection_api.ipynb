# 01 Data Collection - SpaceX API

Collect Falcon 9 launch data using the SpaceX REST API and save raw JSON/CSV.

```python
import requests
import pandas as pd

url = 'https://api.spacexdata.com/v3/launches'
r = requests.get(url)
r.raise_for_status()
data = r.json()
df = pd.json_normalize(data)
df.to_csv('../data/raw/spacex_launches_raw.csv', index=False)
df.head()
```
